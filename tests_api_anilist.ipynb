D'abord une petite fonction qui affiche les dictionnaires et qui filtre le contenu pour adulte

```python
def afficher_animes(reponse):
    collection = reponse.get("data", {}).get("MediaListCollection")

    if not collection:
        print("Aucune collection trouvée.")
        return

    for liste in collection.get("lists", []):
        print(f"\n{liste.get('name', 'Liste sans nom')} :")

        for entree in liste.get("entries", []):
            media = entree.get("media") or {}
            titre = (media.get("title") or {}).get("romaji", "Titre inconnu")
            identifiant = media.get("id")
            afficher_animes(f"- {titre} (ID : {identifiant})")
```

Test de la requête proposé par la doc de l'API graphql d'anilist

```python
import requests
# Here we define our query as a multi-line string
query = '''
query ($id: Int) { # Define which variables will be used in the query (id)
  Media (id: $id, type: ANIME) { # Insert our variables into the query arguments (id) (type: ANIME is hard-coded in the query)
    title {
      romaji
      english
      native
    }
  }
}
'''

# Define our query variables and values that will be used in the query request
variables = {
    'id': 15125
}

url = 'https://graphql.anilist.co'

# Make the HTTP Api request
response = requests.post(url, json={'query': query, 'variables': variables})

print(response.json())
```

Requête paginée

```python
query = '''
query ($id: Int, $page: Int, $perPage: Int, $search: String) {
    Page (page: $page, perPage: $perPage) {
        pageInfo {
            currentPage
            hasNextPage
            perPage
        }
        media (id: $id, search: $search) {
            id
            title {
                romaji
            }
        }
    }
}
'''
variables = {
    'search': 'Fate/Zero',
    'page': 1,
    'perPage': 3
}


response = requests.post(url, json={'query': query, 'variables': variables})
print(response.json())
```
